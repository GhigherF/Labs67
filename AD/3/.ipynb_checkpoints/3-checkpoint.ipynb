```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('Real estate.csv')
corr_matrix = df.corr(numeric_only=True)


plt.figure(figsize=(10, 8))
sns.heatmap(
    corr_matrix, 
    annot=True,         
    fmt=".2f",          
    cmap="coolwarm",    
    vmin=-1, vmax=1,    
    linewidths=0.5,     
    square=True         
)

plt.title("Матрица корреляций признаков", fontsize=14, pad=15)
plt.tight_layout()
plt.show()
```
