```python
# 1. Загрузка данных
df = pd.read_csv("NIFTY_IT.csv")
initial_rows = len(df)
cols_before = df.shape[1]

# Порог пропусков (30%)
threshold = 0.3

# 2. ШАГ 1: Сначала удаляем СТРОКИ с высоким содержанием NaN.
# Это "обрежет" верхний временной блок, где сфокусированы пропуски в 14-м столбце.
min_non_na_per_row = int((1 - threshold) * cols_before)
df_clean = df.dropna(thresh=min_non_na_per_row, axis=0)

dropped_rows = initial_rows - len(df_clean)

# 3. ШАГ 2: Теперь проверяем СТОЛБЦЫ на оставшихся данных.
# Если после "обрезки" проблемных строк в столбцах останутся пропуски >30%, они удалятся.
rows_after_row_drop = len(df_clean)
min_non_na_per_col = int((1 - threshold) * rows_after_row_drop)
df_clean = df_clean.dropna(thresh=min_non_na_per_col, axis=1)

dropped_cols = cols_before - df_clean.shape[1]
final_rows = len(df_clean)

# 4. Вывод статистики
print(f"Исходное количество строк: {initial_rows}")
print(f"Удалено строк (с >{int(threshold*100)}% пропусков): {dropped_rows}")
print(f"Удалено столбцов (с >{int(threshold*100)}% пропусков): {dropped_cols}")
print(f"Итоговое количество строк в очищенном датасете: {final_rows}")

# 5. Визуализация очищенного датасета
plt.figure(figsize=(10, 4))
plt.imshow(df_clean.isnull(), cmap="viridis", aspect="auto", interpolation="nearest")
plt.title("Матрица пропущенных значений (После очистки)", fontsize=12)
plt.xlabel("Индекс столбца")
plt.ylabel("Индекс строки")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import skew
from sklearn.impute import KNNImputer

# 1. Загрузка данных и очистка по вашему сценарию
df = pd.read_csv("NIFTY_IT.csv")
threshold = 0.3

df_clean = df.dropna(
    thresh=int((1 - threshold) * df.shape[1]), axis=0
).copy()
df_clean = df_clean.dropna(
    thresh=int((1 - threshold) * len(df_clean)), axis=1
)

numeric_cols = df_clean.select_dtypes(include=[np.number]).columns
target_col = df_clean[numeric_cols].isna().sum().idxmax()

# 2. Метод 1: Замена на Среднее
df_mean = df_clean.copy()
df_mean[numeric_cols] = df_mean[numeric_cols].fillna(
    df_mean[numeric_cols].mean()
)

# 3. Метод 2: Замена на Медиану
df_median = df_clean.copy()
df_median[numeric_cols] = df_median[numeric_cols].fillna(
    df_median[numeric_cols].median()
)

# 4. Метод 3: KNN Импутация
df_knn = df_clean.copy()
knn_imputer = KNNImputer(n_neighbors=5)
df_knn[numeric_cols] = knn_imputer.fit_transform(df_knn[numeric_cols])

# ---------------------------------------------------------
# 5. Расчёт статистических метрик
# ---------------------------------------------------------


def get_stats(series, name):
    s = series.dropna()
    q25, q75 = s.quantile(0.25), s.quantile(0.75)
    return {
        "Метод": name,
        "Кол-во NaN": series.isna().sum(),
        "Среднее (Mean)": round(s.mean(), 2),
        "Стд. откл. (Std)": round(s.std(), 2),
        "Медиана (Median)": round(s.median(), 2),
        "Межквартильный размах (IQR)": round(q75 - q25, 2),
        "Асимметрия (Skewness)": round(skew(s), 2),
    }


stats_data = [
    get_stats(df_clean[target_col], "Исходные (без NaN)"),
    get_stats(df_mean[target_col], "Замена на Mean"),
    get_stats(df_median[target_col], "Замена на Median"),
    get_stats(df_knn[target_col], "Замена на KNN"),
]

stats_df = pd.DataFrame(stats_data)
print(f"=== Сравнительные метрики для столбца '{target_col}' ===")
print(stats_df.to_string(index=False))

# ---------------------------------------------------------
# 6. Визуализация с наложением метрик (Mean и Median)
# ---------------------------------------------------------
datasets = [
    (df_clean[target_col].dropna(), "1. Исходные данные (без NaN)", "skyblue"),
    (df_mean[target_col], "2. Замена на Среднее (Mean)", "orange"),
    (df_median[target_col], "3. Замена на Медиану (Median)", "green"),
    (df_knn[target_col], "4. Замена методом KNN (k=5)", "purple"),
]

fig, axes = plt.subplots(2, 2, figsize=(14, 9), sharey=True)
axes = axes.flatten()

for idx, (data, title, color) in enumerate(datasets):
    ax = axes[idx]
    ax.hist(data, bins=30, color=color, edgecolor="black", alpha=0.6)

    # Добавляем вертикальные линии для наглядности метрик
    mean_val = data.mean()
    median_val = data.median()

    ax.axvline(
        mean_val,
        color="red",
        linestyle="--",
        linewidth=1.5,
        label=f"Mean: {mean_val:.1f}",
    )
    ax.axvline(
        median_val,
        color="black",
        linestyle="-.",
        linewidth=1.5,
        label=f"Median: {median_val:.1f}",
    )

    ax.set_title(title, fontsize=11, fontweight="bold")
    ax.set_xlabel("Значение")
    if idx % 2 == 0:
        ax.set_ylabel("Частота")
    ax.legend(loc="upper right")

plt.suptitle(
    f"Сравнение распределений и метрик для столбца '{target_col}'",
    fontsize=14,
    y=1.02,
)
plt.tight_layout()
plt.show()
```

```python
# 1. Загрузка данных
df = pd.read_csv("NIFTY_IT.csv")
initial_rows = len(df)

# 2. Подсчет количества полных дубликатов
duplicates_count = df.duplicated().sum()
print(f"Количество найденных дубликатов: {duplicates_count}")

# 3. Удаление дубликатов (сохраняется первая найденная строка)
if duplicates_count > 0:
    df_no_duplicates = df.drop_duplicates().copy()
    print(f"Удалено дубликатов: {duplicates_count}")
else:
    df_no_duplicates = df.copy()
    print("Дубликатов не обнаружено, удаление не требуется.")

final_rows = len(df_no_duplicates)

print(f"Количество строк до удаления дубликатов: {initial_rows}")
print(f"Количество строк после удаления дубликатов: {final_rows}")
```

```python
# 1. Загрузка данных
df = pd.read_csv("NIFTY_IT.csv")
df["Date"] = pd.to_datetime(df["Date"])

# Выбираем основные ценовые столбцы
price_cols = ["Open", "High", "Low", "Close"]


# ---------------------------------------------------------
# График 2: Динамика цены Close с подсветкой аномалий (IQR)
# ---------------------------------------------------------
# Считаем выбросы по Межквартильному размаху (IQR) для цены закрытия
Q1 = df["Close"].quantile(0.25)
Q3 = df["Close"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Фильтр аномальных цен
outliers_close = df[(df["Close"] < lower_bound) | (df["Close"] > upper_bound)]

plt.figure(figsize=(12, 5))
# Линия цены закрытия
plt.plot(
    df["Date"], df["Close"], label="Цена Close", color="navy", linewidth=1.5
)

# Точки выбросов
plt.scatter(
    outliers_close["Date"],
    outliers_close["Close"],
    color="red",
    s=25,
    label=f"Аномалии/Выбросы (Всего: {len(outliers_close)})",
    zorder=5,
)

# Линии границ IQR
plt.axhline(
    upper_bound,
    color="orange",
    linestyle="--",
    label=f"Верхняя граница IQR ({upper_bound:.1f})",
)
plt.axhline(
    lower_bound,
    color="green",
    linestyle="--",
    label=f"Нижняя граница IQR ({lower_bound:.1f})",
)

plt.title("Динамика цены Close с точки зрения IQR-аномалий", fontsize=12)
plt.xlabel("Дата")
plt.ylabel("Цена (INR)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Загрузка данных
df = pd.read_csv("NIFTY_IT.csv")

# 2. Создание текстовых категорий для задания
if "Date" in df.columns:
    df["Date"] = pd.to_datetime(df["Date"])
    df["Day_Name"] = df["Date"].dt.day_name()
    df["Quarter_Name"] = "Q" + df["Date"].dt.quarter.astype(str)
    df = df.drop(columns=["Date"])

df["Trend"] = np.where(df["Daily_Return"] >= 0, "Up", "Down")

# 3. ПОИСК КАТЕГОРИАЛЬНЫХ СТОЛБЦОВ (Исправлено)
cat_cols = df.select_dtypes(
    include=["object", "string", "category"]
).columns.tolist()

print(f"Найденные категориальные столбцы: {cat_cols}")

# --- Label Encoding ---
df_label = df.copy()
for col in cat_cols:
    le = LabelEncoder()
    df_label[col] = le.fit_transform(df_label[col].astype(str))

# --- One-Hot Encoding ---
df_onehot = pd.get_dummies(df, columns=cat_cols, dtype=int)
```
