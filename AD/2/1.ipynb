```python
import pandas as pd

df = pd.read_csv("NIFTY_IT.csv")

missing_count = df.isnull().sum()
print("Количество пропусков по столбцам:\n", missing_count)


missing_percent = (df.isnull().sum() / len(df)) * 100
print("\nПроцент пропусков (%):\n", missing_percent.round(2))

missing_table = pd.DataFrame({
    'Пропуски (шт)': missing_count,
    'Пропуски (%)': missing_percent
})

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 4))

plt.imshow(df.isnull(), cmap='viridis', aspect='auto', interpolation='bilinear')
plt.title('Матрица пропущенных значений', fontsize=12)
plt.xlabel('Индекс столбца')
plt.ylabel('Индекс строки (Время)')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NIFTY_IT.csv")
initial_rows = len(df)


column_threshold = 0.3
cols_before = df.shape[1]
df = df.dropna(thresh=int((1 - column_threshold) * initial_rows), axis=1)
dropped_cols = cols_before - df.shape[1]


row_threshold = 0.3
min_non_na_per_row = int((1 - row_threshold) * df.shape[1])
df_clean = df.dropna(thresh=min_non_na_per_row, axis=0)
final_rows = len(df_clean)
dropped_rows = initial_rows - final_rows
print(f"Исходное количество строк: {initial_rows}")
print(f"Удалено столбцов с >30% пропусков: {dropped_cols}")
print(f"Удалено строк с >30% пропусков: {dropped_rows}")
print(f"Итоговое количество строк в датасете: {final_rows}")





plt.figure(figsize=(10, 4))

plt.imshow(df.isnull(), cmap='viridis', aspect='auto', interpolation='bilinear')
plt.title('Матрица пропущенных значений', fontsize=12)
plt.xlabel('Индекс столбца')
plt.ylabel('Индекс строки (Время)')
plt.show()
```
