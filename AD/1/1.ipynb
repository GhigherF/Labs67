```python
import numpy as np
rnd = np.random.default_rng()
arr = rnd.integers(low=14,high=67,size=(4,5))
arr
```

```python
arr1=arr[:2,:]
print()
print("arr1:")
print(arr1)
arr2=arr[2:4,:]
print()
print("arr2:")
print(arr2)
el1 = arr1 == 18
el2 = arr2 > 28
print()
print(el1)
print()
print(el2)
print()
print(arr1[el1])
print(arr2[el2])
print("count 1:",arr1[el1].size)
print("count 2:",arr2[el2].size)

```
