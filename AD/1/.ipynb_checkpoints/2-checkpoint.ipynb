```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("NIFTY_IT.csv")

df['Date'] = pd.to_datetime(df['Date'])
df['Close'] = pd.to_numeric(df['Close'], errors='coerce')

plt.figure(figsize=(10, 5))
plt.hist(df['Close'].dropna(), bins=40, color='#A61FCC', edgecolor='black', alpha=0.5)
plt.title('Частота нахождения цены NIFTY IT в различных диапазонах (Close)', fontsize=13, fontweight='bold')
plt.xlabel('Цена закрытия (Close)', fontsize=11)
plt.ylabel('Количество дней (Частота)', fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

mean_val = df['Close'].mean()
median_val = df['Close'].median()
print(f"Среднее значение (Mean): {mean_val:.2f}")
print(f"Медиана (Median): {median_val:.2f}\n")

plt.figure(figsize=(7, 6))
plt.boxplot(df['Close'].dropna(), patch_artist=True, 
            boxprops=dict(facecolor='#A61FCC', color='black'),
            medianprops=dict(color='red', linewidth=5))
plt.title('Box Plot (Ящик с усами)', fontsize=13, fontweight='bold')
plt.ylabel('Цена закрытия (Close)', fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

stats = df['Close'].describe()
print(stats)
print("\n" + "="*50 + "\n")

df['Year'] = df['Date'].dt.year

year_a = 2015
year_b = 2021

data_a = df[df['Year'] == year_a]['Close'].dropna()
data_b = df[df['Year'] == year_b]['Close'].dropna()

plt.figure(figsize=(10, 5))
plt.hist(data_a, bins=20, color='#3498db', edgecolor='black', alpha=0.6, label=f'Цены в {year_a} году')
plt.hist(data_b, bins=20, color='#e74c3c', edgecolor='black', alpha=0.6, label=f'Цены в {year_b} году')
plt.title(f'Сравнение распределения цен NIFTY IT ({year_a} vs {year_b})', fontsize=13, fontweight='bold')
plt.xlabel('Цена закрытия (Close)', fontsize=11)
plt.ylabel('Количество дней', fontsize=11)
plt.legend(fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
